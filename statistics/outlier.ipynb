```python
import numpy as np
arr = np.array([2,3,4,6,7,8,9,12,13,16,17,23,27,34,37,201])
arr
```

```python
q1 = np.percentile(arr,25)
q3 = np.percentile(arr,75)
```

```python
q1,q3
```

```python
iqr = q3-q1
iqr
```

```python
uf = q3+1.5*iqr
lf = q1-1.5*iqr
```

```python
l= []
for i in arr:
    if i>=lf and i<=uf:
        l.append(i)
arr2= np.array(l)
```

```python
print(arr)
print(arr2)
```

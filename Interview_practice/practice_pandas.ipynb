### Serializing JSON

```python
import json
marks = {
    'student1':{
        'math':50,
        'english':65,
        'nepali':78
    },
    'student2':{
        'math':50,
        'english':65,
        'nepali':78
    }
}

with open(file='serializable.json',mode='w') as file:
    json.dump(marks,file)
```

### Deserializing JSON

```python
import json
with open(file='serializable.json',mode='r') as read_it:
    data = json.load(read_it)
    print(type(data))
```

### Export Pandas Dataframe to a CSV file

When working on a Data Science project one of the key tasks is data management which includes data collection, cleaning and storage. Once our data is cleaned and processed it's essential to save it in a structured format for further analysis or sharing.

CSV (Comma-Separated Values) file is a widely used format for storing tabular data. Pandas provides to_csv() function to export a DataFrame into a CSV file.

```python
import pandas as pd

scores = {'Name': ['a', 'b', 'c', 'd'],
          'Score': [90, 80, 95, 20]}

df = pd.DataFrame(scores)
print(df)
```

```python
df.to_csv("your_name.csv")
```

#### Customizing the CSV Export 
1. Remove Index Column

```python
df.to_csv('your_name.csv',index=False)
```

2. Export only selected columns 

```python
df.to_csv('your_name.csv',index=False,columns=['Name'])
```

3. Exclude Header Row 

```python
df.to_csv('your_name.csv',index=False,header=False)
```

4. Handling Missing Values 

```python
df.to_csv('your_name.csv',index=False,na_rep ='nothing')
```

5. Change Column Separator 

```python
df.to_csv('your_name.csv',index=False,sep='\t')
```

## Exporting Pandas DataFrame to JSON File

```python
import pandas as pd

df = pd.DataFrame([['a', 'b', 'c'], ['d', 'e', 'f'], ['g', 'h', 'i']],
                  index=['row 1', 'row 2', 'row 3'],
                  columns=['col 1', 'col 2', 'col 3'])
```

```python
df.to_json('file.json',orient='split',compression='infer',index=True)
```

```python
df = pd.read_json('file.json',orient='split',compression='infer')
print(df)
```

## Slicing Pandas Dataframe

```python
import pandas as pd

player_list = [['M.S.Dhoni', 36, 75, 5428000],
               ['A.B.D Villers', 38, 74, 3428000],
               ['V.Kohli', 31, 70, 8428000],
               ['S.Smith', 34, 80, 4428000],
               ['C.Gayle', 40, 100, 4528000],
               ['J.Root', 33, 72, 7028000],
               ['K.Peterson', 42, 85, 2528000]]

df = pd.DataFrame(player_list, columns=['Name', 'Age', 'Weight', 'Salary'])
print(df)
```

#### Slicing Using iloc[] (Integer-Based Indexing)

```python
df.iloc[0:4]
```

```python
df.iloc[:,0:2]
```

```python
df.iloc[2,0]
```

```python
df[df['Age']>35]
```

#### Slicing Using loc[]

```python
df.set_index('Name',inplace=True)
```

```python
df.loc['M.S.Dhoni':'C.Gayle']
```

```python
df.loc['M.S.Dhoni':'C.Gayle','Age':'Salary']
```

```python
df.loc['V.Kohli','Salary']
```

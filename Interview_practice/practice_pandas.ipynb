```python
import pandas as pd
sheet1 = pd.read_excel('students.xlsx',sheet_name=0,index_col=0)
print(sheet1)

sheet2 = pd.read_excel('students.xlsx',sheet_name=1,index_col=0)
print(sheet2)

sheet = pd.concat([sheet1,sheet2])
print(sheet)
```

### Serializing JSON

```python
import json
marks = {
    'student1':{
        'math':50,
        'english':65,
        'nepali':78
    },
    'student2':{
        'math':50,
        'english':65,
        'nepali':78
    }
}

with open(file='serializable.json',mode='w') as file:
    json.dump(marks,file)
```

### Deserializing JSON

```python
import json
with open(file='serializable.json',mode='r') as read_it:
    data = json.load(read_it)
    print(type(data))
```

```python
json_var ="""
{
    "Country": {
        "name": "INDIA",
        "Languages_spoken": [
            {
                "names": ["Hindi", "English", "Bengali", "Telugu"]
            }
        ]
    }
}
"""

json.loads(json_var)
```

```python
import seaborn as sns
import pandas as pd
from scipy.stats import f_oneway
```

```python
df = sns.load_dataset('titanic')
```

```python
df.head()
```

```python
df['pclass'].unique()
```

```python
df = df[['age','pclass']].dropna()
```

is there any relation between pclass and the age

```python
class_1 = df[df['pclass']==1]['age']
class_2 = df[df['pclass']==2]['age']
class_3 = df[df['pclass']==3]['age']
```

```python
f_stats,p_value = f_oneway(class_1,class_2,class_3)
```

```python
f_stats
```

```python
p_value
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print('rejecting the null hypothesis i.e there is a significant difference between at least one passenger class')
else:
    print('acceping the null hypothesis i.e there is no significant difference')
```

```python
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(n_samples=500, n_features=5, centers=3, cluster_std=1.5, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_scaled = scaler.fit_transform(X)
```

```python
x_scaled
```

```python
from sklearn.decomposition import PCA
pca = PCA(n_components = 2)
```

```python
x_transformed = pca.fit_transform(x_scaled)
```

```python
x_transformed
```

```python
df = pd.DataFrame(x_transformed,columns=['pca1','pca2'])
```

```python
# do the k mean clustering on the transformed data 
from sklearn.cluster import DBSCAN
model = DBSCAN(eps=0.3,min_samples = 5)
df['labels']= y
```

```python
sns.scatterplot(x=df['pca1'],y=df['pca2'],hue=df['labels'],palette= 'rainbow')
```
